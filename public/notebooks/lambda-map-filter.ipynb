## 1. Lambda kya hai — aur kya nahi

Lambda ek anonymous, single-expression function hai. Isme `return` keyword nahi likhte aur multiple statements nahi ho sakte. `def` se difference: naam, docstring, statements aur readability.


```python
add = lambda x, y: x + y
print(add(2, 3), (lambda x, y, z: x * y * z)(2, 3, 4))

greet = lambda name: f'Hello {name}'
print(greet('Ahetesham'))

# conditional expression lambda ke andar allowed hai
classify = lambda n: 'even' if n % 2 == 0 else 'odd'
print([classify(n) for n in range(5)])

# def version — jab logic badi ho
def add_def(x, y):
    """Same kaam, naam + docstring ke saath."""
    return x + y

print(add_def(2, 3), add.__name__, add_def.__name__)
```

## 2. sort/sorted me lambda — sabse zyada use hone wala pattern

`key=` isliye hota hai ki comparison value aap decide karo (tuple ka koi element, dict ka field, ya computed value). Secondary key ke liye tuple return karo.


```python
lst = [(3, 2, 3), (4, 5, 6), (1, 2, 3), (2, 6, 7)]
print(sorted(lst, key=lambda x: x))
print(sorted(lst, key=lambda x: x[1]))

people = [
    {'Name': 'Salman', 'Designation': 'Senior Data Engineer', 'Salary': 125000},
    {'Name': 'Rehan', 'Designation': 'Senior Test Engineer', 'Salary': 120000},
    {'Name': 'Anjali', 'Designation': 'Data Analyst', 'Salary': 122000},
]
print([p['Name'] for p in sorted(people, key=lambda p: p['Salary'])])
print([p['Name'] for p in sorted(people, key=lambda p: p['Designation'])])
print(list(map(lambda p: p['Name'], people)))

words = ['banana', 'kiwi', 'apple', 'fig']
print(sorted(words, key=lambda w: (len(w), w)))
```

## 3. map — har element par transform

`map` lazy iterator deta hai, isliye output dekhne ke liye `list()` me wrap karo. Multiple iterables diye ja sakte hain (shortest par ruk jata hai).


```python
nums = [2, 3, 4, 5, 6, 7]
print(list(map(lambda x: x ** 2, nums)))
print(list(map(lambda x: x ** 3, nums)))

l1, l2 = [2, 1, 3, 4, 5], [6, 7, 8, 9, 9]
print(list(map(lambda x, y: x + y, l1, l2)))
print(list(map(lambda x, y: x - y, l1, l2)))

l3 = [20, 21, 22, 23, 25]
print(list(map(lambda a, b, c: a + b + c, l1, l2, l3)))

tuples = [(1, 2, 3), (10, 20, 30), ('HI', 'Hello')]
print(list(map(str, tuples)))
print(list(map(lambda t: [t[0]], tuples[:1])))

print('sum via map:', sum(map(lambda x: x, nums)), '| plain:', sum(nums))
```

## 4. filter — condition se chhaanti

`filter(func, iterable)` sirf woh elements deta hai jinke liye function truthy return kare. `lambda x: x` bhi kaam karta hai (empty hata deta hai).


```python
l1 = [10, 20, 30, 40, 50]
l2 = [5, 6, 7, 8, 9, 10]
print('common  :', list(filter(lambda x: x in l2, l1)))
print('difference:', list(filter(lambda x: x not in l2, l1)))

nums = [5, 6, 7, 8, 9, 10]
print('odd :', list(filter(lambda x: x % 2 != 0, nums)))
print('even:', list(filter(lambda x: x % 2 == 0, nums)))

letters = ['a', 'q', 't', 'e', 'g', 't', 'u']
print('non-vowels:', list(filter(lambda x: x not in 'aeiou', letters)))

vals = [12, 13, 26, 16, 38]
print('div by 13:', list(filter(lambda x: x % 13 == 0, vals)))
print('div by 19:', list(filter(lambda x: x % 19 == 0, vals)))

print('positives sum:', sum(filter(lambda x: x > 0, [-2, -3, -4, 3, 4, 6])))
print('negatives sum:', sum(filter(lambda x: x < 0, [-2, -3, -4, 3, 4, 6])))
```

## 5. String problems lambda ke saath

Palindrome, startswith, isdigit aur max/min length — sab lambda callbacks me translate hote hain.


```python
words = ['level', 'salman', 'impression', 'malayalam']
print('palindromes:', list(filter(lambda w: w == w[::-1], words)))

starts = lambda text, ch: text.lower().startswith(ch)
print(starts('Impression', 'i'), starts('Impression', 'x'))

is_number = lambda s: s.isdigit()
print([(s, is_number(s)) for s in ('12344', '12.5', 'abc')])

longest = ['python', 'machine', 'datascience', 'algorithm']
print('longest :', max(longest, key=len), '| shortest:', min(longest, key=len))

def is_sorted(items):
    return items == sorted(items)

print(is_sorted(['Apple', 'Banana', 'Cat']), is_sorted(['Banana', 'Apple']))

print('tuple intersection:', tuple(filter(lambda x: x in (10, 20, 30, 40, 50, 60), (1, 2, 3, 4, 50, 60))))
```

## 6. reduce aur Fibonacci (wo sawaal jo khali chhoot gaya tha)

`functools.reduce(function, iterable, initial)` do-argument function se poore sequence ko ek value me badal deta hai. Fibonacci ke liye isse accumulator ke roop me use kar sakte hain.


```python
from functools import reduce

nums = [3, 4, 5, 7, 8, 9]
print('product:', reduce(lambda a, b: a * b, nums, 1))
print('max    :', reduce(lambda a, b: a if a > b else b, nums))
print('sum    :', reduce(lambda a, b: a + b, nums, 0))

# Fibonacci: reduce ka accumulator (a, b) tuple hai — har step me (b, a+b)
def fibonacci(n):
    return list(reduce(lambda acc, _: acc + [acc[-1] + acc[-2]], range(n - 2), [0, 1])) if n > 1 else [0][:n]

print('fibonacci 10:', fibonacci(10))

# Lambda + while ka bhi clean version (performance ke liye yehi better hai)
def fib_iter(n):
    a, b, out = 0, 1, []
    for _ in range(n):
        out.append(a)
        a, b = b, a + b
    return out

print('iterative   :', fib_iter(10))
print('multiply each by 3:', list(map(lambda x: x * 3, [3, 4, 5, 7])))
```

## 7. Wo chaar sawaal jo notebook me khali the — ab poore jawab

1. **Anagram grouping** lambda key ke saath: har word ka sorted-letters signature banao.
2. **Binary anagram check**: do numbers ke binary strings ke character counts compare karo.
3. **Stop words remove (filter)**: lambda + set membership.
4. **Same pair count (map)**: `map` se pairs banao, phir ginti.


```python
from collections import Counter
from itertools import groupby

# 1. Anagram groups (groupby + sorted lambda key)
words = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']
groups = groupby(sorted(words, key=lambda w: ''.join(sorted(w))), key=lambda w: ''.join(sorted(w)))
print([list(g) for _, g in groups])

# 2. Binary representations anagram hain?
def binary_anagram(a, b):
    return Counter(bin(a)[2:]) == Counter(bin(b)[2:])

print(binary_anagram(10, 12), binary_anagram(10, 11))   # 1010 vs 1100 / 1011

# 3. Stop words hatao (filter lambda)
STOP = {'is', 'a', 'of', 'and', 'the', 'to', 'in'}
text = 'machine learning is a field of study and it needs data'
print(list(filter(lambda w: w.lower() not in STOP, text.split())))

# 4. Same pair count (map + zip se pairs)
a, b = [1, 2, 3, 4], [1, 9, 3, 8]
pairs = list(map(lambda x, y: (x, y), a, b))
print(pairs, '| same:', sum(map(lambda p: p[0] == p[1], pairs)))
```

## 8. Do classic traps

**Trap 1 (logic, error nahi):** late binding — closures loop variable ko *baad me* dekhte hain, isliye default argument se value bind karni padti hai.

**Trap 2 (`TypeError`):** `map` ke saath `sum()` galat tarike se jodna, ya `reduce` bina initial value ke khali iterable par.


```python
# Trap 1: sab lambdas aakhri i dekhte hain
buggy = [lambda: i for i in range(3)]
print('buggy :', [f() for f in buggy])

fixed = [lambda i=i: i for i in range(3)]
print('fixed :', [f() for f in fixed])

# Wahi bug map ke saath
funcs = list(map(lambda i: (lambda: i), range(3)))
print('map buggy:', [f() for f in funcs])
```

## 10. Interview drills

1. Lambda vs def: performance, readability aur debug-ability ka trade-off?
2. `map`/`filter` vs list comprehension: kaunsa choose karte ho aur kyun (lazy vs eager)?
3. `reduce` kab galat tool hai? Department-wise sum ke liye kya behtar hai?
4. Late-binding closure ko production me kaise pakdogi (test kaise likhoge)?


```python
import sys, timeit

print('map lazy hai (generator jaisa):', map(lambda x: x * 2, range(3)))

eager = [x * 2 for x in range(1000)]
lazy = list(map(lambda x: x * 2, range(1000)))
print('same result:', eager == lazy, '| bytes:', sys.getsizeof(eager))

print('comprehension vs map (µs):')
print('  comprehension:', round(timeit.timeit('[x * 2 for x in range(1000)]', number=200) * 5000, 1))
print('  map          :', round(timeit.timeit('list(map(lambda x: x * 2, range(1000)))', number=200) * 5000, 1))

from collections import defaultdict
rows = [('nlp', 120), ('cv', 80), ('nlp', 90)]
totals = defaultdict(int)
for cat, val in rows:
    totals[cat] += val
print('defaultdict beats reduce here:', dict(totals))
```
