## 1. Basics — for, while, enumerate

`range(start, stop, step)` me **stop include nahi hota**. `for` known collection ke liye, `while` tab jab termination condition data se aati ho.


```python
for i in range(1, 6):
    print(i, end=' ')
print()

for i in range(10, 0, -3):
    print(i, end=' ')
print()

for idx, ch in enumerate('abc'):
    print(idx, ch)

i = 0
while i < 3:
    print('while', i)
    i += 1        # ye line na ho to infinite loop
```

## 2. break, continue, pass

`break` loop chhodta hai, `continue` current iteration skip karta hai, `pass` kuch nahi karta (placeholder). Nested me `break` sirf apne andar wale loop se nikalta hai.


```python
for n in range(1, 10):
    if n == 3:
        continue          # 3 skip
    if n == 7:
        break             # 7 par ruk jao
    print(n, end=' ')
print()

for row in range(3):
    for col in range(5):
        if col == 2:
            break         # inner loop hi rukta hai, outer chalta rehta hai
        print(f'{row}{col}', end=' ')
    print()
```

## 3. Accumulators — sum, product, counting

Har accumulator loop se **pehle** initialize karo. Sum ka identity `0`, product ka `1` hai.


```python
total = 0
for i in range(1, 11, 2):
    total += i
print('odd sum 1..10:', total, '| builtin check:', sum(range(1, 11, 2)))

num = 7
for i in range(1, 11):
    print(f'{num} X {i} = {num * i}')

n = 5
fact = 1
for i in range(2, n + 1):
    fact *= i
print('factorial:', fact)

nums = [2, 4, 1, 6, 8, 7]
even = odd = 0
for x in nums:
    if x % 2 == 0:
        even += 1
    else:
        odd += 1
print('even:', even, 'odd:', odd)
```

## 4. while ke classic problems — digits, reverse, palindrome, Collatz

Digit nikalne ka formula: `n % 10` last digit deta hai, `n // 10` usse hata deta hai.


```python
def count_digits(n):
    count = 0
    while n != 0:
        n //= 10
        count += 1
    return count

def reverse_number(n):
    rev = 0
    while n != 0:
        rev = rev * 10 + n % 10
        n //= 10
    return rev

print(count_digits(786), reverse_number(121))

num = 121
print('palindrome:', num == reverse_number(num))

def collatz_steps(n):
    steps = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps += 1
    return steps

print('collatz(27):', collatz_steps(27))
print('longest under 30:', max(range(1, 30), key=collatz_steps))
```

## 5. for-else — "mila to break, warna not found"

`else` tab chalta hai jab loop bina `break` ke poora ho jaye. Isse flag variable rakhne ki zaroorat nahi padti.


```python
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break
    else:
        return True
    return False

print('primes to 50:', [n for n in range(1, 51) if is_prime(n)])

def find_user(users, name):
    for user in users:
        if user == name:
            print('found:', name)
            break
    else:
        print('not found:', name)

find_user(['Neha', 'Ali'], 'Ali')
find_user(['Neha', 'Ali'], 'Rahul')

for i in []:
    pass
else:
    print('khaali loop par bhi else chalta hai')
```

## 6. Patterns — nested loops ka asli use

Outer loop rows deta hai, inner loop us row ke characters. `print(..., end=' ')` same line me rakhta hai, khaali `print()` line todta hai.


```python
rows = 5

print('--- triangle ---')
for i in range(1, rows + 1):
    print('* ' * i)

print('--- pyramid ---')
for i in range(1, rows + 1):
    print(' ' * (rows - i) + '* ' * i)

print('--- numbers ---')
count = 1
for i in range(1, rows + 1):
    for j in range(i):
        print(count, end=' ')
        count += 1
    print()

print('--- alphabets ---')
char = ord('A')
for i in range(1, rows + 1):
    print(' '.join(chr(char + j) for j in range(i)))
    char += i
```

## 8. Practice set

1. 1..n ka sum bina formula, phir formula `n*(n+1)//2` se — dono compare karo.
2. First n prime numbers nikalo (count-based `while`).
3. Fibonacci ke 10 terms aur n-th term dono.
4. Armstrong number check karo (407, 9474).
5. Range me divisibility lists banao (3, 5, 7 ke liye alag list).
6. Multiplication table grid 1..5 aur har row ka sum.
7. Retry loop: max 3 attempts, success par break, warna else me give-up message.


```python
# 1
n = 100
loop_sum = 0
for i in range(1, n + 1):
    loop_sum += i
print('loop vs formula:', loop_sum, n * (n + 1) // 2)

# 2
def first_primes(count):
    found, candidate = [], 2
    while len(found) < count:
        for d in range(2, int(candidate ** 0.5) + 1):
            if candidate % d == 0:
                break
        else:
            found.append(candidate)
        candidate += 1
    return found

print('first 8 primes:', first_primes(8))

# 3
a, b = 0, 1
series = []
for _ in range(10):
    series.append(a)
    a, b = b, a + b
print('fibonacci:', series)

# 4
def is_armstrong(num):
    digits = str(num)
    return num == sum(int(d) ** len(digits) for d in digits)

print('armstrong:', [n for n in (407, 9474, 123) if is_armstrong(n)])

# 5
by_3, by_5, by_7 = [], [], []
for num in range(1, 51):
    if num % 3 == 0:
        by_3.append(num)
    if num % 5 == 0:
        by_5.append(num)
    if num % 7 == 0:
        by_7.append(num)
print(by_3[:5], by_5[:5], by_7[:5])

# 6
for i in range(1, 6):
    row = [i * j for j in range(1, 6)]
    print(row, 'sum =', sum(row))

# 7
attempts, max_attempts = 0, 3
while attempts < max_attempts:
    attempts += 1
    ok = attempts == 3          # simulate: 3rd try succeeds
    print(f'attempt {attempts}: {"ok" if ok else "retry"}')
    if ok:
        break
else:
    print('gave up after', max_attempts, 'attempts')
```

## 9. Interview drills

1. Two-pointer palindrome check (`while left < right`) — O(1) extra space.
2. Sliding window ka window-maximum: naive O(n·k) vs deque O(n).
3. Nested loop ko set se O(n) me convert karo (do lists ka common).
4. Ek infinite loop ko production-safe kaise banate ho? (max attempts + timeout + logging)


```python
from collections import deque
import time

def is_palindrome(s):
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

print('palindrome:', is_palindrome('malayalam'), is_palindrome('python'))

def window_max_naive(nums, k):
    return [max(nums[i:i + k]) for i in range(len(nums) - k + 1)]

def window_max_fast(nums, k):
    dq, out = deque(), []
    for i, n in enumerate(nums):
        while dq and nums[dq[-1]] < n:
            dq.pop()
        dq.append(i)
        if dq[0] <= i - k:
            dq.popleft()
        if i >= k - 1:
            out.append(nums[dq[0]])
    return out

nums = [1, 3, -1, -3, 5, 3, 6, 7]
print('naive:', window_max_naive(nums, 3))
print('deque:', window_max_fast(nums, 3))

def common_fast(a, b):
    lookup = set(b)
    return [x for x in a if x in lookup]

print('common O(n):', common_fast([1, 2, 3, 40], [40, 5, 3]))

max_wait, start = 0.05, time.monotonic()
checks = 0
while True:
    checks += 1
    if time.monotonic() - start >= max_wait:
        print('timeout guard hit after', checks, 'checks')
        break
```
