## 2. Adding and removing elements

`append` end me O(1) hai, `insert` beech me O(n) kyunki baaki elements shift hote hain. `remove` value se hatata hai (na mile to **ValueError**), `pop` index se (**IndexError** agar khali ho).


```python
marks = [78, 92, 45]
marks.append(88)
marks.insert(1, 61)
print(marks)

marks.remove(45)          # value se
popped = marks.pop()      # end se, value return karta hai
print(marks, 'popped:', popped)

print(marks.pop(0), marks)
marks.clear()
print('after clear:', marks)
```

## 3. sort vs sorted, reverse vs reversed

`sort()` / `reverse()` **in-place** hain aur `None` return karte hain. `sorted()` / `reversed()` naya object dete hain. Isi wajah se `x = lst.sort()` ka bug hota hai (x None ho jata hai).


```python
nums = [78, 92, 45, 88, 61]
print('sorted() copy :', sorted(nums))
print('original      :', nums)

nums.sort(reverse=True)
print('sort() in-place:', nums)

students = [('Ali', 88), ('Neha', 92), ('Raj', 75)]
print(sorted(students, key=lambda s: s[1]))
print(sorted(students, key=lambda s: -s[1])[:2])

print(list(reversed(nums)))
```

## 4. Class jaisa sawaal: min/max aur sort **bina built-in** ke

Interview me yahi pattern poocha jata hai — loop se best/worst track karo, phir built-in se verify karo.


```python
nums = [1, 12, 3, 14, 25]

biggest = smallest = nums[0]
for n in nums[1:]:
    if n > biggest:
        biggest = n
    if n < smallest:      # do alag if — elif se ek branch skip ho jati hai
        smallest = n
print(biggest, smallest, '| builtin:', max(nums), min(nums))

ascending = [10, 12, 123, 45, 34]
for i in range(len(ascending)):
    for j in range(i + 1, len(ascending)):
        if ascending[i] > ascending[j]:
            ascending[i], ascending[j] = ascending[j], ascending[i]
print('manual sort:', ascending)
```

## 5. List comprehension aur aggregates

Comprehension = map + filter ek line me. Product ke liye identity `1` se shuru karo (sum ka `0` hota hai).


```python
import math

nums = [1, 2, 3, 4, 5, 6, 7]
print('sum', sum(nums), '| product', math.prod(nums))
print('even count', sum(1 for n in nums if n % 2 == 0), '| odd count', sum(n % 2 for n in nums))

squares_of_evens = [n ** 2 for n in nums if n % 2 == 0]
square_or_cube = [n ** 2 if n % 2 == 0 else n ** 3 for n in nums]
print(squares_of_evens)
print(square_or_cube)

sentence = 'We are working on the jupyter notebook'
print([w for w in sentence.split() if len(w) > 5])
```

## 6. Duplicates aur common elements

`set()` order tod deta hai. Order chahiye to `dict.fromkeys()` ya seen-set pattern use karo. Common elements ke liye set intersection O(n+m) hai (nested loop O(n·m)).


```python
nums = [1, 2, 3, 4, 45, 55, 6, 7, 3, 4]
print('order preserved:', list(dict.fromkeys(nums)))
print('sorted unique   :', sorted(set(nums)))

l1 = [1, 2, 3, 40, 50, 69]
l2 = [10, 20, 30, 40, 50, 69]
print('common :', [x for x in l1 if x in set(l2)])
print('only l1:', [x for x in l1 if x not in set(l2)])
```

## 7. Nested lists — flatten, sublist detection, type split

Flatten me `isinstance(x, (list, tuple))` check karo: `extend` sirf sequence par chalta hai, aur string ko bhi sequence maan liya jaye to characters alag ho jate hain.


```python
nested = [1, [2, 3], [4, [5, 6]], 7]
flat = []
for item in nested:
    if isinstance(item, list):
        for inner in item:
            if isinstance(inner, list):
                flat.extend(inner)
            else:
                flat.append(inner)
    else:
        flat.append(item)
print('flattened:', flat)

mixed = [1, 2, 3.5, 2.5, 'a', 'hello', True]
ints = [x for x in mixed if isinstance(x, int) and not isinstance(x, bool)]
floats = [x for x in mixed if isinstance(x, float)]
strings = [x for x in mixed if isinstance(x, str)]
print(ints, floats, strings)

sublists = [x for x in [1, [2, 3], 4] if isinstance(x, list)]
print('has sublist:', bool(sublists), sublists)
```

## 8. Pairing, rotating aur zipping

`zip` do lists ko pair karta hai (chhoti list par ruk jata hai). Rotation ke liye slicing sabse readable hai: `lst[-k:] + lst[:-k]`.


```python
a = [2, 3, 4, 5, 6]
b = [10, 20, 30, 40, 50]
print('product :', [x * y for x, y in zip(a, b)])
print('pairs   :', list(zip(a, b))[:2])

nums = [1, 2, 3, 4, 5, 6]
k = 2
print('rotated :', nums[-k:] + nums[:-k])

print('alternate:', [x for pair in zip([1, 2, 3], ['a', 'b', 'c']) for x in pair])
```

## 9. Aliasing vs copy (sabse common silent bug)

`b = a` sirf naya naam deta hai, nayi list nahi. Nested data me `copy()` bhi kaafi nahi hoti — `deepcopy` chahiye.


```python
from copy import deepcopy

original = [[1, 2], [3, 4]]
alias = original
shallow = original.copy()
deep = deepcopy(original)

alias[0].append(99)      # alias = original isliye dono badalte hain
print('original:', original)
print('shallow (inner list shared):', shallow)
print('deep (independent):', deep)
```

## 11. Practice set — khud likho, phir niche ke cells se compare karo

1. `[11, 33, 50]` ko `113350` integer banao (bina `join` ka shortcut sochkar).
2. Third-largest number nikalo duplicates ke saath (set + sort).
3. Do lists ka element-wise difference nikalo.
4. List palindrome hai ya nahi check karo.
5. 1..50 me se prime numbers ki list banao.
6. Har element ka square karo, sirf 50 se bade rakho (comprehension).
7. List ko 4-4 ke chunks me todo.
8. `k` steps left rotate karo (extra list ke bina sochna).


```python
# 1
lst = [11, 33, 50]
print(int(''.join(str(i) for i in lst)))

# 2
scores = [12, 41, 45, 45, 23, 15]
print('third largest:', sorted(set(scores), reverse=True)[2])

# 3
l1, l2 = [12, 13, 14, 15], [4, 5, 6, 7]
print('diff:', [x - y for x, y in zip(l1, l2)])

# 4
pal = [1, 2, 3, 2, 1]
print('palindrome:', pal == pal[::-1])

# 5
def is_prime(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True

print('primes:', [n for n in range(1, 51) if is_prime(n)])

# 6
print('big squares:', [n ** 2 for n in range(1, 11) if n ** 2 > 50])

# 7
items = list(range(1, 11))
print('chunks:', [items[i:i + 4] for i in range(0, len(items), 4)])

# 8
def rotate_left(lst, k):
    k %= len(lst)
    return lst[k:] + lst[:k]

print('left rotate:', rotate_left([1, 2, 3, 4, 5], 2))
print('k > len   :', rotate_left([1, 2, 3], 5))
```

## 12. Interview drills

1. Top-3 largest nikaalo bina poora sort kiye (`heapq.nlargest` vs manual heap ka complexity batao).
2. Do sorted lists ko merge karo O(n+m) me aur duplicate count do.
3. `list.pop(0)` slow kyun hai? FIFO ke liye kya use karoge?
4. List aur tuple me kab kya choose karte ho — ek real example do.


```python
import heapq
from collections import deque

nums = [5, 1, 9, 3, 7, 2, 8]
print('top-3 :', heapq.nlargest(3, nums))
print('bottom-2:', heapq.nsmallest(2, nums))

def merge_sorted(a, b):
    i = j = dupes = 0
    out = []
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            out.append(a[i]); i += 1
        elif a[i] > b[j]:
            out.append(b[j]); j += 1
        else:
            out += [a[i], b[j]]
            dupes += 1
            i += 1; j += 1
    return out + a[i:] + b[j:], dupes

print('merge :', merge_sorted([1, 3, 5, 7], [2, 3, 5, 8]))

queue = deque(['build', 'test'])
queue.append('deploy')
print('deque popleft (O(1)):', queue.popleft(), list(queue))
```
